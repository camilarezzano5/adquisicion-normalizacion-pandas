```python

import pandas as pd

# ARCHIVO 1: TRANSACCIONES (CSV)

transacciones = {
    "id_producto": [
        101, 102, 103, 104, 105,
        101, 102, 103, 106, 107,
        108, 101, 104, 105, 106,
        102, 103, 107, 108, 104
    ],
    "cantidad": [
        2, 3, 1, 4, 2,
        1, 5, 2, 3, 1,
        2, 3, 1, 4, 2,
        2, 3, 2, 1, 5
    ],
    "fecha": [
        "01/10/2026", "02/10/2026", "03/10/2026",
        "04/10/2026", "05/10/2026", "06/10/2026",
        "07/10/2026", "08/10/2026", "09/10/2026",
        "10/10/2026", "11/10/2026", "12/10/2026",
        "13/10/2026", "14/10/2026", "15/10/2026",
        "16/10/2026", "17/10/2026", "18/10/2026",
        "19/10/2026", "20/10/2026"
    ]
}

df_transacciones = pd.DataFrame(transacciones)

# Guardar CSV
df_transacciones.to_csv(
    "transacciones.csv",
    index=False
)

# ARCHIVO 2: PRODUCTOS (EXCEL)

productos = {
    "id_producto": [101, 102, 103, 104, 105, 106, 107, 108],
    "nombre_producto": [
        "Notebook", "Mouse", "Teclado", "Monitor",
        "Auriculares", "Impresora", "Webcam", "Parlantes"
    ],
    "precio_unitario": [
        850000, 15000, 25000, 180000,
        45000, 120000, 35000, 28000
    ]
}

df_productos = pd.DataFrame(productos)

# Guardar Excel
df_productos.to_excel(
    "productos.xlsx",
    index=False
)

print("¡Archivos creados correctamente!")

display(df_transacciones.head())
display(df_productos)
```

```python

# PASO 2: ADQUISICION DE DATOS

import pandas as pd

# Cargar archivo CSV
transacciones = pd.read_csv("transacciones.csv")

# Cargar archivo Excel
productos = pd.read_excel("productos.xlsx", engine="openpyxl")

# Mostrar las primeras filas
print("DATOS DE TRANSACCIONES:")
display(transacciones.head())

print("CATALOGO DE PRODUCTOS:")
display(productos.head())

# Verificar dimensiones
print("Dimensiones de transacciones:", transacciones.shape)
print("Dimensiones de productos:", productos.shape)
```

```python

# PASO 3: LIMPIEZA Y NORMALIZACION

# 1. Verificar tipos de datos
print("Tipos de datos originales:")
print(transacciones.dtypes)
print(productos.dtypes)

# 2. Convertir identificadores a tipo entero
transacciones["id_producto"] = pd.to_numeric(
    transacciones["id_producto"], errors="coerce"
)

productos["id_producto"] = pd.to_numeric(
    productos["id_producto"], errors="coerce"
)

# 3. Convertir cantidad y precio a valores numericos
transacciones["cantidad"] = pd.to_numeric(
    transacciones["cantidad"], errors="coerce"
)

productos["precio_unitario"] = pd.to_numeric(
    productos["precio_unitario"], errors="coerce"
)

# 4. Convertir fecha a formato datetime
transacciones["fecha"] = pd.to_datetime(
    transacciones["fecha"],
    format="%d/%m/%Y",
    errors="coerce"
)

# 5. Eliminar valores nulos en campos criticos
transacciones = transacciones.dropna(
    subset=["id_producto", "cantidad", "fecha"]
)

productos = productos.dropna(
    subset=["id_producto", "nombre_producto", "precio_unitario"]
)

# 6. Convertir identificadores a enteros
transacciones["id_producto"] = transacciones["id_producto"].astype(int)
productos["id_producto"] = productos["id_producto"].astype(int)

# 7. Eliminar productos duplicados
productos = productos.drop_duplicates(
    subset=["id_producto"]
)

# 8. Mostrar resultados
print("\nTipos de datos despues de la limpieza:")
print(transacciones.dtypes)
print(productos.dtypes)

print("\nValores nulos en transacciones:")
print(transacciones.isnull().sum())

print("\nValores nulos en productos:")
print(productos.isnull().sum())

print("\nTransacciones limpias:")
display(transacciones.head())

print("\nProductos limpios:")
display(productos)
```

```python

# PASO 4: INTEGRACION DE DATOS

# Unir transacciones con el catalogo de productos
ventas = pd.merge(
    transacciones,
    productos,
    on="id_producto",
    how="inner"
)

# Mostrar el resultado
print("DATASET INTEGRADO:")
display(ventas)

# Verificar dimensiones
print("Cantidad de filas y columnas:", ventas.shape)

# Verificar si quedaron valores nulos
print("\nValores nulos:")
print(ventas.isnull().sum())
```

```python

# PASO 5: CREACION DE LA COLUMNA CALCULADA

# Calcular el total de cada venta
ventas["total_venta"] = (
    ventas["cantidad"] * ventas["precio_unitario"]
)

# Mostrar el dataset actualizado
print("DATASET CON TOTAL DE VENTAS:")
display(ventas)

# Calcular el total general vendido
print("Total general de ventas: $", ventas["total_venta"].sum())

# Verificar valores nulos
print("\nValores nulos:")
print(ventas.isnull().sum())

# Verificar tipos de datos
print("\nTipos de datos:")
print(ventas.dtypes)
```

```python

# PASO 6: EXPORTACION A PARQUET

# Instalar PyArrow
!pip install pyarrow -q

# Exportar dataset
ventas.to_parquet(
    "ventas_final.parquet",
    index=False,
    engine="pyarrow"
)

# Leer nuevamente el archivo para comprobarlo
ventas_parquet = pd.read_parquet("ventas_final.parquet")

print("¡Archivo Parquet generado correctamente!")

display(ventas_parquet.head())

print("Dimensiones del archivo:", ventas_parquet.shape)
```

```python

import os
import shutil
import zipfile

# Crear estructura del proyecto
os.makedirs("proyecto_pandas/datos", exist_ok=True)

# Copiar archivos originales
shutil.copy("transacciones.csv", "proyecto_pandas/datos/transacciones.csv")
shutil.copy("productos.xlsx", "proyecto_pandas/datos/productos.xlsx")

# Copiar resultado Parquet
shutil.copy("ventas_final.parquet", "proyecto_pandas/ventas_final.parquet")

# Crear requirements.txt
with open("proyecto_pandas/requirements.txt", "w") as f:
    f.write("pandas\nopenpyxl\npyarrow\n")

# Crear README
readme = """
# Adquisición y Normalización con Pandas

## Descripción
Proyecto de integración de datos provenientes de dos fuentes:
CSV de transacciones y Excel de productos.

## Librerías
- Pandas
- Openpyxl
- PyArrow

## Ejecución
1. Instalar dependencias:
   pip install -r requirements.txt

2. Ejecutar:
   python adquisicion.py

## Proceso realizado
- Carga de CSV y Excel.
- Limpieza de datos.
- Conversión de fechas a datetime.
- Tratamiento de valores nulos.
- Integración mediante merge.
- Cálculo de total_venta.
- Exportación a Parquet.

## Resultado
Archivo final: ventas_final.parquet
"""

with open("proyecto_pandas/README.md", "w", encoding="utf-8") as f:
    f.write(readme)

# Crear archivo ZIP
with zipfile.ZipFile("proyecto_pandas.zip", "w", zipfile.ZIP_DEFLATED) as zipf:
    for root, dirs, files in os.walk("proyecto_pandas"):
        for file in files:
            ruta = os.path.join(root, file)
            zipf.write(ruta, ruta)

print("Proyecto preparado correctamente.")
```
